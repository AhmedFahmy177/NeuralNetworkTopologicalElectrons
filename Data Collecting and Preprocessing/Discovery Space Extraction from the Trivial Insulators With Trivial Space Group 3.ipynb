```python
import pandas as pd
import ast 

# Extracting the space  group number to use it later when constructing the Discovery vs Labelled datasets, as discussed in the presentation vedio in detail.
df = pd.read_csv("materiae_Trivial_Ins_withSOC.csv")


def extract_spacegroup_number(s):
    try:
       
        d = ast.literal_eval(s)
        return d.get('number', None)
    except Exception:
        return None 


df['spacegroup_number'] = df['spacegroup'].apply(extract_spacegroup_number)


df.to_csv("materiae_Trivial_Ins_withSOC_withSGnum.csv", index=False)

print(df[['spacegroup', 'spacegroup_number']].head())
```
