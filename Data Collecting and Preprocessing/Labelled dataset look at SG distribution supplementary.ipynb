```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("materiae_All_Labelled_withSOC_with_Composition_Vectors.csv")


assert "spacegroup_number" in df.columns, "Column 'spacegroup_number' not found in dataset."

spacegroup_counts = df["spacegroup_number"].value_counts().sort_index()

print("Total materials:", len(df))
print("\nDistribution (spacegroup_number → count):\n")
print(spacegroup_counts)

plt.figure(figsize=(14,5))
plt.bar(spacegroup_counts.index, spacegroup_counts.values, color='steelblue')
plt.xlabel("Space Group Number")
plt.ylabel("Number of Materials")
plt.title("Distribution of Materials by Space Group Number")
plt.tight_layout()
plt.show()
```
