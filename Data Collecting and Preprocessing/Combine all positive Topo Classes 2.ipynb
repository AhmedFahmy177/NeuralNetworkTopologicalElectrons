```python
import pandas as pd

# Combining the different imported topological class files into one
files = {
    "materiae_TIwithSOC.csv": "TI",
    "materiae_TCIwithSOC.csv": "TCI",
    "materiae_HSPSMwithSOC.csv": "HSPSM",
    "materiae_HSLSMwithSOC.csv": "HSLSM"
}

dfs = []

for file, label in files.items():
    df = pd.read_csv(file)
    df["topological_class"] = label 
    dfs.append(df)


combined = pd.concat(dfs, ignore_index=True)


combined.to_csv("materiae_AllTopological_withSOC.csv", index=False)

print(f"✅ Combined {len(dfs)} files → total {len(combined)} materials saved to materiae_AllTopological_withSOC.csv")
```
