```python
import pandas as pd

# Load both datasets 
file1 = "topogivity_found_materials_withCompositionVectors.csv"
file2 = "materiae_filtered_by_spacegroups_Trivial_Insulators_Trivial_SI_withCompositionVectors.csv"

df_topogivity = pd.read_csv(file1)
df_trivial = pd.read_csv(file2)

if "formula" not in df_topogivity.columns or "formula" not in df_trivial.columns:
    raise ValueError("❌ One of the files does not contain a 'formula' column!")

df_topogivity["formula_clean"] = df_topogivity["formula"].str.replace(r"\s+", "", regex=True).str.strip()
df_trivial["formula_clean"] = df_trivial["formula"].str.replace(r"\s+", "", regex=True).str.strip()

intersection = set(df_topogivity["formula_clean"]) & set(df_trivial["formula_clean"])
intersection_list = sorted(list(intersection))

output_file = "intersecting_materials_topogivity_vs_trivial.csv"
pd.DataFrame({"formula": intersection_list}).to_csv(output_file, index=False)

print(f"✅ Found {len(intersection_list)} intersecting materials.")
print(f"Results saved to '{output_file}'")

print("\n🔍 Intersecting materials:")
print(intersection_list[:20])  # print first 20 for inspection
```

```python
import pandas as pd

topogivity_file = "topogivity_found_materials_withCompositionVectors.csv"
intersect_file = "intersecting_materials_topogivity_vs_labelled.csv"

df_topo = pd.read_csv(topogivity_file)
df_intersect = pd.read_csv(intersect_file)

df_topo["formula_clean"] = df_topo["formula"].str.replace(r"\s+", "", regex=True).str.strip()
df_intersect["formula_clean"] = df_intersect["formula"].str.replace(r"\s+", "", regex=True).str.strip()

filtered_df = df_topo[~df_topo["formula_clean"].isin(df_intersect["formula_clean"])].copy()

filtered_df.drop(columns=["formula_clean"], inplace=True)

output_file = "topogivity_found_materials_without_intersections.csv"
filtered_df.to_csv(output_file, index=False)

print(f"✅ Removed {len(df_topo) - len(filtered_df)} intersecting materials.")
print(f"Remaining materials: {len(filtered_df)}")
print(f"Cleaned dataset saved as '{output_file}'")
```
